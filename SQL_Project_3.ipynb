```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
import sqlite3
```

```python
file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)
```

```python
conn = sqlite3.connect(':memory:')
```

```python
df.to_sql('sales_data', conn, index=False,
          if_exists='replace')
print("Database Created Successfully")
```

```python
query = "SELECT * FROM sales_data"
result = pd.read_sql(query, conn)
result.head()
```

```python
query = """
SELECT Product, TotalPrice
FROM sales_data
"""
result = pd.read_sql(query, conn)
result.head()
```

```python
print(f"Maximum TotalPrice in DataFrame: {df['TotalPrice'].max()}")
query = """
SELECT *
FROM sales_data
WHERE TotalPrice > 3000
"""
result = pd.read_sql(query, conn)
result.head()
```

```python
query = """
SELECT Product, TotalPrice
FROM sales_data
ORDER BY TotalPrice DESC
"""
result = pd.read_sql(query, conn)
result.head()
```

```python
query = """
SELECT PaymentMethod, COUNT(*) as Total_Orders
FROM sales_data
GROUP BY PaymentMethod
"""
result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT SUM(TotalPrice) as Total_Revenue
FROM sales_data
"""
result = pd.read_sql(query, conn)
result
```

```python
query = """
SELECT AVG(TotalPrice) as Average_Price
FROM sales_data
"""
result = pd.read_sql(query, conn)
result
```
